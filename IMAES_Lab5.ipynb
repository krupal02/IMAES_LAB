```python
# ============================================
# Experiment 5: Backward Chaining
# Car Fault Diagnosis
# ============================================

# Each rule has:
# Rule ID
# Conclusion / Hypothesis
# Conditions required for that conclusion

rules = [
    {
        "id": "R1",
        "conclusion": "battery fault",
        "conditions": ["dim headlights", "weak horn"]
    },

    {
        "id": "R2",
        "conclusion": "alternator fault",
        "conditions": ["battery warning light", "battery keeps losing charge"]
    },

    {
        "id": "R3",
        "conclusion": "starter motor fault",
        "conditions": ["clicking sound", "engine not cranking"]
    },

    {
        "id": "R4",
        "conclusion": "fuel system fault",
        "conditions": ["engine cranks", "engine does not start"]
    }
]
```

```python
# Store facts that have already been determined
known_facts = {}

# Store the rule currently being used
current_rule = {}

def ask_fact(fact):
    """
    Ask the user whether a fact is true.
    """

    if fact in known_facts:
        return known_facts[fact]

    while True:

        answer = input(
            f"Is '{fact}' true? (yes/no/why): "
        ).strip().lower()

        if answer in ["yes", "y"]:
            known_facts[fact] = True
            return True

        elif answer in ["no", "n"]:
            known_facts[fact] = False
            return False

        elif answer == "why":
            print("WHY:")
            print(current_rule.get("reason",
                  "This fact is required to test the current hypothesis."))

        else:
            print("Please enter yes, no, or why.")
```

```python
def backward_chain(goal, rules):
    """
    Try to prove a goal using backward chaining.
    """

    global current_rule

    # If the goal is already known
    if goal in known_facts:
        return known_facts[goal]

    # Find rules that can conclude the goal
    matching_rules = [
        rule for rule in rules
        if rule["conclusion"] == goal
    ]

    # If there is no rule for this goal,
    # ask the user directly.
    if not matching_rules:
        return ask_fact(goal)

    # Try each rule
    for rule in matching_rules:

        current_rule = {
            "reason": (
                f"{rule['id']} is being checked because "
                f"it can establish '{rule['conclusion']}'. "
                f"It requires: {', '.join(rule['conditions'])}."
            )
        }

        print(f"\nTrying {rule['id']}:")
        print(
            f"IF {' AND '.join(rule['conditions'])} "
            f"THEN {rule['conclusion']}"
        )

        all_conditions_true = True

        # Prove every condition
        for condition in rule["conditions"]:

            if not backward_chain(condition, rules):
                all_conditions_true = False
                break

        if all_conditions_true:
            known_facts[goal] = True
            return True

    known_facts[goal] = False
    return False
```

```python
hypotheses = [
    "battery fault",
    "alternator fault",
    "starter motor fault",
    "fuel system fault"
]
```

```python
def diagnose(hypotheses):

    global known_facts

    # Reset facts for a new diagnosis
    known_facts = {}

    print("=" * 50)
    print("CAR FAULT DIAGNOSIS")
    print("=" * 50)

    for hypothesis in hypotheses:

        print(f"\nChecking hypothesis: {hypothesis}")

        if backward_chain(hypothesis, rules):

            print("\n" + "=" * 50)
            print("FAULT IDENTIFIED:", hypothesis.upper())
            print("=" * 50)

            return hypothesis

    print("\n" + "=" * 50)
    print("No fault could be identified.")
    print("=" * 50)

    return None
```
