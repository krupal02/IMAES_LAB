```python
# ============================================
# FUZZY LOGIC - STUDENT PERFORMANCE
# ============================================

def attendance_membership(attendance):

    # Low attendance
    low = max(0, min(1, (60 - attendance) / 30))

    # Medium attendance
    if attendance <= 60:
        medium = attendance / 60
    elif attendance <= 80:
        medium = (80 - attendance) / 20
    else:
        medium = 0

    # High attendance
    high = max(0, min(1, (attendance - 70) / 30))

    return {
        "low": low,
        "medium": medium,
        "high": high
    }


def marks_membership(marks):

    # Low marks
    low = max(0, min(1, (50 - marks) / 50))

    # Medium marks
    if marks <= 50:
        medium = marks / 50
    elif marks <= 80:
        medium = (80 - marks) / 30
    else:
        medium = 0

    # High marks
    high = max(0, min(1, (marks - 70) / 30))

    return {
        "low": low,
        "medium": medium,
        "high": high
    }
```

```python
fuzzy_rules = [
    ("low", "low", "poor"),
    ("low", "medium", "average"),
    ("low", "high", "good"),

    ("medium", "low", "average"),
    ("medium", "medium", "average"),
    ("medium", "high", "good"),

    ("high", "low", "average"),
    ("high", "medium", "good"),
    ("high", "high", "excellent")
]
```

```python
def calculate_performance(attendance, marks):

    attendance_values = attendance_membership(attendance)
    marks_values = marks_membership(marks)

    performance_values = {
        "poor": 0,
        "average": 0,
        "good": 0,
        "excellent": 0
    }

    # Apply fuzzy rules
    for att_level, mark_level, performance in fuzzy_rules:

        # AND operation = minimum
        strength = min(
            attendance_values[att_level],
            marks_values[mark_level]
        )

        # Aggregate rules using maximum
        performance_values[performance] = max(
            performance_values[performance],
            strength
        )

    return attendance_values, marks_values, performance_values
```

```python
def defuzzify(performance_values):

    output_values = {
        "poor": 25,
        "average": 50,
        "good": 75,
        "excellent": 100
    }

    numerator = 0
    denominator = 0

    for category, membership in performance_values.items():

        numerator += membership * output_values[category]
        denominator += membership

    if denominator == 0:
        return 0

    return numerator / denominator
```

```python
def student_performance(attendance, marks):

    attendance_values, marks_values, performance_values = \
        calculate_performance(attendance, marks)

    score = defuzzify(performance_values)

    if score < 40:
        label = "Poor"
    elif score < 60:
        label = "Average"
    elif score < 80:
        label = "Good"
    else:
        label = "Excellent"

    print("=" * 50)

    print(f"Attendance: {attendance}%")
    print(f"Marks: {marks}%")

    print("\nAttendance Membership:")
    for key, value in attendance_values.items():
        print(f"{key}: {value:.2f}")

    print("\nMarks Membership:")
    for key, value in marks_values.items():
        print(f"{key}: {value:.2f}")

    print("\nPerformance Membership:")
    for key, value in performance_values.items():
        print(f"{key}: {value:.2f}")

    print(f"\nFinal Performance Score: {score:.2f}")
    print(f"Performance: {label}")

    print("=" * 50)

    return score, label
```

```python
# ============================================
# CONSTRAINT SATISFACTION - TIMETABLE
# ============================================

subjects = [
    "AI",
    "ML",
    "DBMS",
    "OS"
]

time_slots = [
    "9-10",
    "10-11",
    "11-12",
    "12-1"
]
```

```python
def is_valid(assignment):

    # Constraint 1:
    # No two subjects can have the same slot

    slots = list(assignment.values())

    if len(slots) != len(set(slots)):
        return False


    # Constraint 2:
    # AI and ML cannot be consecutive

    if "AI" in assignment and "ML" in assignment:

        ai_index = time_slots.index(assignment["AI"])
        ml_index = time_slots.index(assignment["ML"])

        if abs(ai_index - ml_index) == 1:
            return False


    # Constraint 3:
    # DBMS cannot be at 12-1

    if "DBMS" in assignment:

        if assignment["DBMS"] == "12-1":
            return False


    return True
```

```python
def solve_timetable(assignment={}):

    # All subjects assigned
    if len(assignment) == len(subjects):
        return assignment.copy()

    # Select next unassigned subject
    subject = next(
        s for s in subjects
        if s not in assignment
    )

    # Try every possible time slot
    for slot in time_slots:

        assignment[subject] = slot

        # Check constraints
        if is_valid(assignment):

            result = solve_timetable(assignment)

            if result is not None:
                return result

        # Undo assignment
        del assignment[subject]

    return None
```

```python
solution = solve_timetable()

print("Generated Timetable")
print("=" * 30)

if solution:

    for subject, slot in solution.items():
        print(f"{subject:10} -> {slot}")

else:

    print("No valid timetable found.")
```

```python
print("\nConstraint Check:")

if solution and is_valid(solution):
    print("All constraints satisfied.")
else:
    print("Constraints violated.")
```
