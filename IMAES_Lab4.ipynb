```python
# Forward Chaining Inference Engine
# Experiment 4 - Zookeeper Animal Identification

# Rules are represented as:
# (Rule ID, conditions, conclusion)

rules = [
    ("R1", ["has hair"], "is mammal"),
    ("R2", ["gives milk"], "is mammal"),
    ("R3", ["has feathers"], "is bird"),
    ("R4", ["is mammal", "eats meat"], "is carnivore"),
    ("R5", ["is mammal", "has pointed teeth"], "is carnivore"),
    ("R6", ["is carnivore", "has tawny color", "has black spots"], "is tiger"),
    ("R7", ["is mammal", "has long neck", "has long legs"], "is giraffe"),
    ("R8", ["is mammal", "has black stripes", "not carnivore"], "is zebra"),
]
```

```python
def forward_chaining(initial_facts, rules):
    # Working memory
    facts = set(initial_facts)

    # Store the order in which rules are fired
    firing_sequence = []

    while True:
        conflict_set = []

        # Find all applicable rules
        for rule_id, conditions, conclusion in rules:

            # Rule is applicable if:
            # 1. All conditions are present
            # 2. Conclusion is not already present
            if all(condition in facts for condition in conditions):
                if conclusion not in facts:
                    conflict_set.append(
                        (rule_id, conditions, conclusion)
                    )

        # No applicable rule -> stop
        if not conflict_set:
            break

        # Select the first rule
        selected_rule = conflict_set[0]

        rule_id, conditions, conclusion = selected_rule

        # Fire the rule
        facts.add(conclusion)

        # Record rule
        firing_sequence.append(rule_id)

    return facts, firing_sequence
```

```python
initial_facts = {
    "has hair",
    "eats meat",
    "has pointed teeth",
    "has tawny color",
    "has black spots"
}

facts, sequence = forward_chaining(initial_facts, rules)

print("Initial Facts:")
for fact in initial_facts:
    print("-", fact)

print("\nFiring Sequence:")
print(" -> ".join(sequence))

print("\nFinal Working Memory:")
for fact in facts:
    print("-", fact)

print("\nIdentification:")

if "is tiger" in facts:
    print("Tiger")
elif "is giraffe" in facts:
    print("Giraffe")
elif "is zebra" in facts:
    print("Zebra")
else:
    print("No specific animal identified.")
```

```python
#Exercise 3: Fire ALL applicable rules
def forward_chaining_all(initial_facts, rules):

    facts = set(initial_facts)

    firing_sequence = []

    cycle = 1

    while True:

        applicable_rules = []

        # Find all applicable rules
        for rule_id, conditions, conclusion in rules:

            if all(condition in facts for condition in conditions):
                if conclusion not in facts:
                    applicable_rules.append(
                        (rule_id, conditions, conclusion)
                    )

        # Stop if no rule can fire
        if not applicable_rules:
            break

        print(f"\nCycle {cycle}")

        # Fire ALL applicable rules
        for rule_id, conditions, conclusion in applicable_rules:

            print(f"Firing {rule_id} -> {conclusion}")

            facts.add(conclusion)
            firing_sequence.append(rule_id)

        cycle += 1

    return facts, firing_sequence
```

```python
initial_facts = {
    "has hair",
    "eats meat",
    "has pointed teeth",
    "has tawny color",
    "has black spots"
}

facts, sequence = forward_chaining_all(
    initial_facts,
    rules
)

print("\nFinal Working Memory:")
for fact in facts:
    print("-", fact)

print("\nFiring Sequence:")
print(" -> ".join(sequence))
```
