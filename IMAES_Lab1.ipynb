# Experiment 1
# Knowledge Representation - Semantic Network with Inheritance

```python
# 1. Semantic Network

knowledge = [
    ("animal", "has", "skin"),

    ("bird", "is-a", "animal"),
    ("bird", "can", "fly"),

    ("fish", "is-a", "animal"),
    ("fish", "can", "swim"),

    ("shark", "is-a", "fish"),
    ("shark", "can", "swim"),

    ("penguin", "is-a", "bird"),
    ("penguin", "cannot", "fly")
]
```

```python
# 2. Find parent of a node

def get_parent(node):
    for subject, relation, obj in knowledge:
        if subject == node and relation == "is-a":
            return obj

    return None
```

```python
# 3. Find inheritance chain

def inheritance_chain(node):
    chain = [node]

    current = node

    while True:
        parent = get_parent(current)

        if parent is None:
            break

        chain.append(parent)
        current = parent

    return chain
```

```python
# 4. Query using inheritance and exceptions

def query(node, relation, value):

    current = node

    while current is not None:

        # First check for exception
        for subject, rel, obj in knowledge:

            if subject == current and rel == "cannot" and obj == value:
                return "cannot " + value

        # Then check for normal property
        for subject, rel, obj in knowledge:

            if subject == current and rel == relation and obj == value:
                return value

        # Move to parent
        current = get_parent(current)

    return "unknown"
```

```python
# 5. Display knowledge base

print("SEMANTIC NETWORK")
print("-" * 40)

for triple in knowledge:
    print(triple)
```

```python
# 6. Display inheritance chains

print("\nINHERITANCE CHAINS")
print("-" * 40)

for node in ["penguin", "shark", "fish", "bird"]:
    print(node, "->", " -> ".join(inheritance_chain(node)))
```

```python
# 7. Queries

print("\nQUERY RESULTS")
print("-" * 40)

queries = [
    ("penguin", "can", "fly"),
    ("bird", "can", "fly"),
    ("shark", "can", "swim"),
    ("penguin", "has", "skin"),
    ("fish", "has", "skin"),
    ("shark", "can", "fly")
]

for node, relation, value in queries:
    result = query(node, relation, value)
    print(f"Query: {node} {relation} {value} -> {result}")
```

#Exercise 2 — Re-implement using Frames

```python
# Frame-based Knowledge Representation

frames = {

    "animal": {
        "has": "skin"
    },

    "bird": {
        "ako": "animal",
        "can": "fly"
    },

    "fish": {
        "ako": "animal",
        "can": "swim"
    },

    "shark": {
        "ako": "fish",
        "can": "swim"
    },

    "penguin": {
        "ako": "bird",
        "cannot": "fly"
    },

    "ostrich": {
        "ako": "bird",
        "cannot": "fly",
        "can": "run-fast"
    }
}
```

```python
# Find inheritance chain

def frame_chain(node):

    chain = [node]
    current = node

    while "ako" in frames[current]:

        parent = frames[current]["ako"]
        chain.append(parent)
        current = parent

    return chain
```

```python
# Query frame knowledge

def frame_query(node, slot, value):

    current = node

    while current is not None:

        frame = frames[current]

        # Check exception first
        if "cannot" in frame and frame["cannot"] == value:
            return "cannot " + value

        # Check requested slot
        if slot in frame and frame[slot] == value:
            return value

        # Move to parent
        current = frame.get("ako")

    return "unknown"
```

```python
# Display frames

print("FRAME REPRESENTATION")
print("=" * 40)

for name, frame in frames.items():

    print("\nFrame:", name)

    for slot, filler in frame.items():
        print("   ", slot, "->", filler)
```

```python
# Display inheritance

print("\nINHERITANCE CHAINS")
print("=" * 40)

for node in frames:
    print(node, "->", " -> ".join(frame_chain(node)))
```

```python
# Test queries

print("\nQUERY RESULTS")
print("=" * 40)

print("Penguin can fly:",
      frame_query("penguin", "can", "fly"))

print("Penguin has skin:",
      frame_query("penguin", "has", "skin"))

print("Ostrich can fly:",
      frame_query("ostrich", "can", "fly"))

print("Ostrich can run-fast:",
      frame_query("ostrich", "can", "run-fast"))

print("Shark can swim:",
      frame_query("shark", "can", "swim"))
```

#Exercise 3 - Multiple Parents

```python
# Multiple Inheritance

multi_frames = {

    "animal": {
        "has": "skin"
    },

    "bird": {
        "ako": ["animal"],
        "can": "fly"
    },

    "aquatic-animal": {
        "ako": ["animal"],
        "can": "swim"
    },

    "duck": {
        "ako": ["bird", "aquatic-animal"]
    }
}
```

```python
# Query with multiple parents

def multi_query(node, slot, value):

    visited = set()

    def search(current):

        if current in visited:
            return None

        visited.add(current)

        frame = multi_frames[current]

        # Check exception first
        if frame.get("cannot") == value:
            return "cannot " + value

        # Check current frame
        if frame.get(slot) == value:
            return value

        # Search parents from left to right
        parents = frame.get("ako", [])

        for parent in parents:

            result = search(parent)

            if result is not None:
                return result

        return None

    result = search(node)

    if result is None:
        return "unknown"

    return result
```

```python
# Test multiple inheritance

print("MULTIPLE INHERITANCE")
print("=" * 40)

print("Duck can fly:",
      multi_query("duck", "can", "fly"))

print("Duck can swim:",
      multi_query("duck", "can", "swim"))

print("Duck has skin:",
      multi_query("duck", "has", "skin"))
```
